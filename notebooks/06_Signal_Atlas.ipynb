# Experiment 06 — Synthetic MIN Signal Atlas

Characterize the action of several causal memory kernels on elementary deterministic signals. This is a descriptive experiment, not a claim that the output should resemble the input.

```python
from pathlib import Path
import importlib.util
ROOT = Path.cwd()
if not (ROOT / 'min').is_dir(): ROOT = ROOT.parent
spec = importlib.util.spec_from_file_location('signal_atlas', ROOT / 'experiments' / '06_signal_atlas.py')
module = importlib.util.module_from_spec(spec)
spec.loader.exec_module(module)
signals = module.build_signals()
kernel_names = ['identity', 'exponential_tau_0.10', 'soe_two_scale', 'powerlaw']
```

```python
rows = []
for signal in signals:
    for kernel_name in kernel_names:
        y = module.apply_memory(signal, kernel_name)
        rows.append((signal.signal_type, kernel_name, module.energy(signal.samples), module.energy(y), module.correlation(signal.samples, y), module.delay_of_max_correlation(signal.samples, y)))
for row in rows:
    print(row)
```

## Interpretation

Treat the output as an atlas of transformations. The identity case is the control. Exponential, finite-SOE, and power-law cases expose different memory timescale structures. The next experiment should characterize frequency response and then move to BPSK/QPSK/QAM signals.
