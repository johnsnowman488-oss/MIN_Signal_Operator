# MIN operator — first numerical experiment

Apply the reference causal MIN operator to a constant signal and compare it with the analytical exponential-kernel result.

```python
import numpy as np
import matplotlib.pyplot as plt
from min import MemoryOperator, exponential_kernel
```

```python
t = np.linspace(0, 5, 1001)
tau = 0.8
x = np.ones_like(t)
y = MemoryOperator(lambda lag: exponential_kernel(lag, tau)).apply(t, x)
exact = tau * (1 - np.exp(-t / tau))
print('max error:', np.max(np.abs(y - exact)))
```

```python
plt.plot(t, y, label='MIN reference')
plt.plot(t, exact, '--', label='analytic')
plt.xlabel('t')
plt.ylabel('output')
plt.legend()
plt.show()
```
