# Experiment 09 — Digital signals through MIN memory
BPSK, QPSK and 16-QAM; EVM after scalar gain removal.

```python
import importlib.util
from pathlib import Path
ROOT=Path.cwd()
while not (ROOT/'experiments'/'09_digital_memory_transform.py').exists(): ROOT=ROOT.parent
spec=importlib.util.spec_from_file_location('experiment09',ROOT/'experiments'/'09_digital_memory_transform.py')
mod=importlib.util.module_from_spec(spec); spec.loader.exec_module(mod)
```

```python
rows=[]
for seed in range(5):
    for name,generator in mod.SIGNALS.items(): rows.extend(mod.run_one(name,generator,seed))
import numpy as np
for signal in mod.SIGNALS:
    for kernel in mod.KERNELS:
        v=[r['evm_percent'] for r in rows if r['signal']==signal and r['kernel']==kernel]
        print(signal,kernel,'mean EVM % =',np.mean(v),'std =',np.std(v,ddof=1))
```

## Interpretation
Large residual EVM after scalar gain removal indicates temporal distortion/inter-symbol memory rather than a simple amplitude/phase change. This is a characterization experiment, not a BER-improvement claim.
