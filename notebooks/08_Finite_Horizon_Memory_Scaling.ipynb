# Experiment 08 — Finite-Horizon Memory Scaling

Measure how quickly finite-history MIN response approaches its infinite-horizon response.

```python
import importlib.util
from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np

ROOT = Path.cwd()
while not (ROOT / 'experiments' / '08_finite_horizon_memory_scaling.py').exists():
    ROOT = ROOT.parent
spec = importlib.util.spec_from_file_location('experiment08', ROOT / 'experiments' / '08_finite_horizon_memory_scaling.py')
mod = importlib.util.module_from_spec(spec)
spec.loader.exec_module(mod)
```

```python
rows = []
for kernel in mod.KERNELS:
    rows.extend(mod.rows_for_kernel(kernel))
summary = mod.summarize(rows)
summary
```

## Horizon-error curves

The plots use relative complex error $E_T(\\omega)$.

```python
for kernel in mod.KERNELS:
    plt.figure(figsize=(7, 4))
    for omega in mod.OMEGAS:
        subset = sorted((r for r in rows if r['kernel'] == kernel.name and r['omega_rad_s'] == float(omega)), key=lambda r: r['horizon_s'])
        plt.loglog([r['horizon_s'] for r in subset], [r['relative_complex_error'] for r in subset], marker='o', label=f'omega={omega:g}')
    plt.title(kernel.name)
    plt.xlabel('Observation horizon T (s)')
    plt.ylabel('Relative complex error')
    plt.grid(True, which='both', alpha=0.25)
    plt.legend()
    plt.show()
```

```python
for kernel in mod.KERNELS:
    print(kernel.name)
    thresholds = summary[kernel.name]['frequency_specific_horizon_thresholds']
    for omega, values in thresholds.items():
        print(f'  omega={omega}: {values["T_for_10pct_or_better_s"]} s for <=10%')
```
