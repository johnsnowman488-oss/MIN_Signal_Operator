# Prony identification and SOE realization

This experiment generates a known finite sum of decaying exponentials, identifies it with the classical Prony construction, and compares the resulting streaming SOE realization with direct MIN quadrature.

```python
import numpy as np
import matplotlib.pyplot as plt
from min import MemoryOperator, SOEMemory, fit_prony
```

```python
dt = 0.01
t = np.arange(80) * dt
weights = np.array([1.7, 0.45])
gammas = np.array([2.0, 11.0])
kernel_samples = sum(w * np.exp(-g * t) for w, g in zip(weights, gammas))
fit = fit_prony(kernel_samples, dt, order=2)
order = np.argsort(fit.gammas.real)
print('true gammas:', gammas)
print('Prony gammas:', fit.gammas.real[order])
print('true weights:', weights)
print('Prony weights:', fit.weights.real[order])
```

```python
x = np.sin(2*np.pi*1.5*t)
direct = MemoryOperator(lambda lag: sum(w*np.exp(-g*lag) for w,g in zip(weights,gammas))).apply(t, x)
streaming = SOEMemory(weights, gammas).apply(t, x)
print('max direct/SOE difference:', np.max(np.abs(direct-streaming)))
plt.plot(t, direct, label='direct MIN quadrature')
plt.plot(t, streaming, '--', label='SOE realization')
plt.legend()
plt.xlabel('t')
plt.ylabel('output')
plt.show()
```
